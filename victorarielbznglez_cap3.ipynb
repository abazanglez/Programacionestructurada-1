```python
def repeat(word, n):
    print(word * n)
```

```python
spam = 'Spam, '
repeat(spam, 4)
```

```python
def first_two_lines():
    repeat(spam, 4)
    repeat(spam, 4)
```

```python
first_two_lines()
```

```python
def last_three_lines():
    repeat(spam, 2)
    print('(Lovely Spam, Wonderful Spam!)')
    repeat(spam, 2)
```

```python
def print_verse():
    first_two_lines()
    last_three_lines()
```

```python
Spam, Spam, Spam, Spam, 
Spam, Spam, Spam, Spam, 
Spam, Spam, 
(Lovely Spam, Wonderful Spam!)
Spam, Spam, 
```

```python
for i in range(2):
    print(i)
```

```python
for i in range(2):
    print("Verse", i)
    print_verse()
    print()
```

```python
def print_n_verses(n):
    for i in range(n):
        print_verse()
        print()
```
